```python
import pandas as pd
```

**1.Read the CSV File**



```python
sales_data = pd.read_csv("/pandas_dataset.csv")
original_shape = sales_data.shape
print(sales_data.head())
```

**2.Rename Columns**

```python
sales_data.columns = sales_data.columns.str.lower().str.replace(" ", "_")
```

**3.Drop Unnecessary Rows**

```python
sales_data = sales_data.iloc[1:].reset_index(drop=True)
```

**4.Drop Columns**

```python
sales_data =sales_data.drop(columns=["product"])
```

**5.Count Missing Values**

```python
print(sales_data.isnull().sum())
```

**6.Handle Missing Values**

a.Drop Rows Based on a Column

```python
sales_data = sales_data.dropna(subset=["sales"])
```

b.Fill Missing Values

```python
sales_data["profit"] =sales_data["profit"].fillna(sales_data["profit"].mean())
sales_data["discount"] = sales_data["discount"].fillna(sales_data["discount"].median())
```

**7.Handle Missing Categorical Data**

```python
sales_data["customer_name"] =sales_data["customer_name"].fillna("Unknown")
```

**8.Detect and Remove Duplicates**

```python
duplicate_count =sales_data.duplicated(subset=["order_id"]).sum()
print(duplicate_count)
before_count = len(sales_data)
sales_data = sales_data.drop_duplicates(subset=["order_id"])
sales_data =sales_data.reset_index(drop=True)
removed_count =before_count - len(sales_data)
print(removed_count)
```

**9.Filtering and creating new columns**

```python
costly_orders =sales_data[sales_data["unit_price"]>20000]
print(costly_orders)

done_orders = sales_data[(sales_data["unit_price"]>10000) & (sales_data["status"]=="Completed")]
print(done_orders[["customer_name","category","status"]])

sales_data["total_amount"] = sales_data["quantity"] *sales_data["unit_price"] - sales_data["discount"]
sales_data["customer_type"] =sales_data["quantity"].apply(lambda qty:"Bulk Buyer" if qty>=3 else "Regular Buyer")
```

**10.Final Clean Dataset**

```python
print("Before:", original_shape, "After:", sales_data.shape)
print(sales_data.head(10))
```
